```python
##configuring logging

import logging

logging.basicConfig(
    filename="app.log",
    filemode="w",
    level=logging.DEBUG,
    format='%(asctime)s=%(name)s=%(levelname)s=%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S',
    force=True
)

logging.debug("this is debug message")
logging.info("this is an info msg")
logging.warning("this is a warning message")
logging.error("this is the error message")
logging.critical("this is a critical message")


```

```python
##logging with multiple loggers
##create a logger for module 
logger1=logging.getLogger("module1")
logger1.setLevel(logging.DEBUG)
##create a logger for module 2 
logger2=logging.getLogger("module2")
logger2.setLevel(logging.WARNING)
##basic configuration for logging
logging.basicConfig(
    filename="app.log",
    filemode="w",
    level=logging.DEBUG,
    format='%(asctime)s=%(name)s=%(levelname)s=%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S',
    force=True
)
##log message with different loggers
logger1.debug("this is debug msg for module 1")
logger2.warning(" this is warning for module 3")

logger2.error(" this is a error message ")

```
