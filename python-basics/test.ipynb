```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Read the CSV file
df = pd.read_csv("filtered_road_accidents.csv")

# 2. Display the data
print("First 5 rows:")
print(df.head())

print("\nDataset information:")
print(df.info())

# 3. Check for missing values
print("\nMissing values:")
print(df.isnull().sum())

# 4. Calculate total accidents for each year
total_2021 = df["Accidents_2021"].sum()
total_2022 = df["Accidents_2022"].sum()
total_2023 = df["Accidents_2023"].sum()
total_2024 = df["Accidents_2024"].sum()

print("\nTotal accidents by year:")
print("2021:", total_2021)
print("2022:", total_2022)
print("2023:", total_2023)
print("2024:", total_2024)

# 5. Store yearly totals in a list
years = [2021, 2022, 2023, 2024]
totals = [total_2021, total_2022, total_2023, total_2024]

# Find the year with the highest total
highest_index = np.argmax(totals)

print("\nYear with the highest accidents:", years[highest_index])
print("Highest total:", totals[highest_index])

# 6. Find the top 10 states in 2024
top10 = df.sort_values(
    "Accidents_2024", ascending=False
).head(10)

print("\nTop 10 states in 2024:")
print(top10[["State_UT", "Accidents_2024"]])

# 7. Calculate percentage change from 2021 to 2024
df["Percentage_Change"] = (
    (df["Accidents_2024"] - df["Accidents_2021"])
    / df["Accidents_2021"]
) * 100

print("\nPercentage change by state:")
print(df[["State_UT", "Percentage_Change"]])

# Find the state with the highest percentage increase
highest_change = df["Percentage_Change"].idxmax()

print("\nHighest percentage increase:")
print(df.loc[highest_change, "State_UT"])
print(df.loc[highest_change, "Percentage_Change"])

# 8. Find Karnataka's accident data
karnataka = df[df["State_UT"] == "Karnataka"]

print("\nKarnataka accident data:")
print(karnataka[
    ["Accidents_2021", "Accidents_2022",
     "Accidents_2023", "Accidents_2024"]
])

# 9. Bar chart: Total accidents by year
plt.bar(years, totals)
plt.title("Total Road Accidents by Year")
plt.xlabel("Year")
plt.ylabel("Number of Accidents")
plt.show()

# 10. Bar chart: Top 10 states in 2024
plt.bar(top10["State_UT"], top10["Accidents_2024"])
plt.title("Top 10 States by Road Accidents in 2024")
plt.xlabel("State / UT")
plt.ylabel("Number of Accidents")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# 11. Line chart: Karnataka's accident trend
karnataka_totals = [
    int(karnataka["Accidents_2021"].iloc[0]),
    int(karnataka["Accidents_2022"].iloc[0]),
    int(karnataka["Accidents_2023"].iloc[0]),
    int(karnataka["Accidents_2024"].iloc[0])
]

plt.plot(years, karnataka_totals, marker="o")
plt.title("Road Accidents in Karnataka")
plt.xlabel("Year")
plt.ylabel("Number of Accidents")
plt.show()

# 12. Bar chart: Highest percentage increases
top_changes = df.sort_values(
    "Percentage_Change", ascending=False
).head(10)

plt.bar(
    top_changes["State_UT"],
    top_changes["Percentage_Change"]
)
plt.title("Top 10 States by Percentage Increase")
plt.xlabel("State / UT")
plt.ylabel("Percentage Change (%)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd 
data=pd.read_csv("filtered_road_accidents.csv")
data=pd.DataFrame(data)

data.head(5)
year=["2021","2020","2023","2024"]
totals=[
    data["Accidents_2021"].sum(),
     data["Accidents_2022"].sum(),
      data["Accidents_2023"].sum(),
       data["Accidents_2024"].sum(),

]
plt.plot(year,totals,marker="o",color="red")
plt.title("Road Accident by Year")
plt.xlabel("Year")
plt.ylabel("Total Acciedent")

plt.show()
top10=data.sort_values(
    "Accidents_2024",ascending=False
).head(5)
print(top10)
plt.bar(top10["State_UT"],top10['Accidents_2024'])
plt.xlabel("State / UT ")
plt.ylabel("Number of Accidents")
plt.xticks(rotation=45)
plt.show()
data["Accident_Change"]=(data["Accidents_2024"]-data["Accidents_2021"])
top_increase=data.sort_values(
    "Accident_Change",ascending=False
    
).head(5)
plt.bar(
    top_increase["State_UT"],
    top_increase["Accident_Change"]
)
plt.title("Top 10 States with the Largest Accident Increase")
plt.xlabel("State / UT")
plt.ylabel("Change in Accidents")
plt.xticks(rotation=45)
plt.show()
```

```python
##seaborn
import seaborn as sns
import matplotlib.pyplot as plt 
##basic plotting with seaborn 
tips=sns.load_dataset('tips')
print(tips)
sns.scatterplot(x="total_bill",y='tip',data=tips)
plt.title("Scatter plot of total bill vs tip ")
plt.show()
##line plot 
sns.lineplot(x="size",y="total_bill",data=tips)
plt.title("line plot of total bill of size ")
plt.show()
##categorical variable plots 
##bar plot
sns.barplot(x="sex",y="total_bill",data=tips)
plt.title("bar plot of total bill by day")
plt.show()
##box plot 
sns.boxplot(x="day",y="total_bill",data=tips)
plt.title("box plot of total bill buy day")
plt.show()
##violinplot
sns.violinplot(x="day",y="total_bill",data=tips)
plt.title("box plot of total bill buy day")
plt.show()
#histogram
sns.histplot(tips["total_bill"],bins=10,kde=True)
plt.title("box plot of total bill buy day")
plt.show()
##kde plot 
sns.kdeplot(tips['total_bill'],fill=True)
##pairplot
sns.pairplot(tips)
plt.show()
##heatmap
corr=tips[['total_bill',"tip","size"]].corr()
sns.heatmap(corr,annot=True,cmap='coolwarm')
plt.show()

```
